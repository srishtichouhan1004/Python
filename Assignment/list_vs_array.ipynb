```python
import numpy as np
import sys

print("NumPy version:", np.__version__)
```

```python
# Python list containing numbers from 1 to 1000
python_list = list(range(1, 1001))

# NumPy array containing numbers from 1 to 1000
numpy_array = np.array(range(1, 1001))

print("Python List:")
print(python_list)

print("\nNumPy Array:")
print(numpy_array)
```

```python
# Memory usage of Python list
list_memory = sys.getsizeof(python_list)

# Memory usage of NumPy array
array_memory = numpy_array.nbytes

print("Memory Usage:")
print("Python List:", list_memory, "bytes")
print("NumPy Array:", array_memory, "bytes")
```

```python
import time
```

```python
def compare_addition_speed():

    # Create 10,000 integers
    python_list = list(range(1, 10001))
    numpy_array = np.array(range(1, 10001))

    # Python List addition
    start_time = time.time()

    list_result = []

    for num in python_list:
        list_result.append(num + 5)

    end_time = time.time()

    list_time = end_time - start_time

    # NumPy Array addition
    start_time = time.time()

    array_result = numpy_array + 5

    end_time = time.time()

    array_time = end_time - start_time

    print("Python List Time:", list_time, "seconds")
    print("NumPy Array Time:", array_time, "seconds")


compare_addition_speed()
```

```python
ratings = np.array([3.5, 4.0, 4.5, 3.8, 4.2])

print("Original Ratings:")
print(ratings)
```

```python
new_ratings_loop = []

for rating in ratings:
    new_ratings_loop.append(rating * 2)

print("Ratings after multiplying by 2 using for-loop:")
print(new_ratings_loop)
```

```python
new_ratings_vectorized = ratings * 2

print("Ratings after multiplying by 2 using NumPy:")
print(new_ratings_vectorized)
```

```python
#Vectorization:
Vectorization in NumPy allows us to perform operations on all elements of an array without using an explicit for loop. 
In the Zomato example, instead of looping through every restaurant rating, we can simply use ratings *
2. NumPy applies the operation to every element automatically. Vectorized operations are usually faster and make the code shorter and easier to read.
```
